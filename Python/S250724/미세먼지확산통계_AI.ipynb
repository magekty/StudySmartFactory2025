```python
# --- 1. 분석에 필요한 라이브러리 불러오기 ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Matplotlib에서 한글 폰트가 깨지는 것을 방지하기 위한 설정입니다.
# 사용하시는 운영체제에 맞춰 아래 코드의 주석을 해제하고 실행해주세요.

# Windows 사용자
from matplotlib import font_manager, rc
font_path = "C:/Windows/Fonts/malgun.ttf" # 맑은 고딕
font = font_manager.FontProperties(fname=font_path).get_name()
rc('font', family=font)

# Mac 사용자
# plt.rcParams['font.family'] = 'AppleGothic'
# plt.rcParams['axes.unicode_minus'] = False # 마이너스 부호 깨짐 방지


# --- 2. 미세먼지 데이터 불러오기 ---
# 가지고 계신 데이터 파일의 실제 경로와 이름으로 수정해주세요.
file_path = 'AIR_HOUR_2021.csv' 

try:
    # CSV 파일을 읽어 DataFrame으로 만듭니다. 한글 파일명/경로 때문에 오류가 나면 encoding='cp949'를 시도해보세요.
    df = pd.read_csv(file_path, encoding='euc-kr')
    print(f"'{file_path}' 파일을 성공적으로 불러왔습니다.")
except FileNotFoundError:
    print(f"'{file_path}' 파일을 찾을 수 없습니다. 파일 경로를 확인해주세요.")
    print("분석을 계속하기 위해 예시 데이터를 생성합니다.")
    # 파일이 없을 경우를 대비한 예시 데이터
    data = {
        '측정일시': ['2023-10-01', '2023-10-02', '2023-10-03', '2023-10-04', '2023-10-05'],
        '지역': ['서울', '부산', '서울', '부산', '서울'],
        '미세먼지(PM10)': [35, 28, 45, 33, 50],
        '초미세먼지(PM2.5)': [20, 15, 28, 18, 32]
    }
    df = pd.DataFrame(data)


# --- 3. 데이터 기본 정보 확인 및 전처리 ---
print("\n[데이터 상위 5개 행 미리보기]")
print(df.head())

print("\n[데이터 기본 정보]")
df.info()

# '측정일시' 컬럼을 문자열(object)에서 날짜/시간(datetime) 타입으로 변환합니다.
# 이렇게 해야 시간 순서대로 정렬하거나 특정 기간의 데이터를 쉽게 추출할 수 있습니다.
if '측정일시' in df.columns:
    df['측정일시'] = pd.to_datetime(df['측정일시'])
    print("\n'측정일시' 컬럼을 datetime 타입으로 변환했습니다.")

# 결측치(비어있는 값)가 있는지 확인합니다.
print("\n[컬럼별 결측치 개수]")
print(df.isnull().sum())

# --- 4. 데이터 시각화를 통한 통계 분석 ---

# 4-1. 시간에 따른 미세먼지/초미세먼지 농도 변화
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='측정일시', y='pm10', label='pm10')
sns.lineplot(data=df, x='측정일시', y='pm2.5', label='pm2.5')
plt.title('시간에 따른 미세먼지 농도 변화 추이')
plt.xlabel('측정일시')
plt.ylabel('농도 (µg/m³)')
plt.grid(True)
plt.legend()
plt.show()


# 4-2. 지역별 평균 미세먼지 농도 비교
if '지역' in df.columns:
    plt.figure(figsize=(10, 6))
    # 지역별로 그룹화하여 평균 미세먼지 농도를 계산합니다.
    avg_by_region = df.groupby('지역')['pm10'].mean().sort_values(ascending=False)
    
    sns.barplot(x=avg_by_region.index, y=avg_by_region.values)
    plt.title('지역별 평균 미세먼지(PM10) 농도 비교')
    plt.xlabel('지역')
    plt.ylabel('평균 농도 (µg/m³)')
    plt.show()

```

```python
file_path = 'AIR_HOUR_2021.csv' 

try:
    # CSV 파일을 읽어 DataFrame으로 만듭니다. 한글 파일명/경로 때문에 오류가 나면 encoding='cp949'를 시도해보세요.
    df = pd.read_csv(file_path, encoding='utf-8')
    print(f"'{file_path}' 파일을 성공적으로 불러왔습니다.")
except FileNotFoundError:
    print(f"'{file_path}' 파일을 찾을 수 없습니다. 파일 경로를 확인해주세요.")
    print("분석을 계속하기 위해 여러 달에 걸친 예시 데이터를 생성합니다.")
    # 파일이 없을 경우를 대비한 예시 데이터 (여러 달)
    data = {
        '측정일시': [
            '2023-10-01', '2023-10-15', '2023-10-30',
            '2023-11-05', '2023-11-12', '2023-11-28',
            '2023-12-10', '2023-12-20', '2023-12-25'
        ],
        '지역': ['서울', '부산', '서울', '부산', '서울', '부산', '서울', '부산', '서울'],
        '미세먼지(PM10)': [35, 28, 45, 60, 55, 65, 80, 75, 88],
        '초미세먼지(PM2.5)': [20, 15, 28, 40, 35, 42, 55, 50, 60]
    }
    df = pd.DataFrame(data)


# --- 3. 데이터 기본 정보 확인 및 전처리 ---
print("\n[데이터 상위 5개 행 미리보기]")
print(df.head())

# '측정일시' 컬럼을 문자열(object)에서 날짜/시간(datetime) 타입으로 변환합니다.
if '측정일시' in df.columns:
    df['측정일시'] = pd.to_datetime(df['측정일시'])
    print("\n'측정일시' 컬럼을 datetime 타입으로 변환했습니다.")

# --- 여기서부터 요청하신 기능이 추가됩니다 ---

# --- 5. 월별 평균 미세먼지 농도 분석 및 시각화 (추가 기능 1) ---
print("\n[월별 평균 미세먼지 농도 분석]")

# '측정일시'에서 '월' 정보를 추출하여 새로운 컬럼을 만듭니다.
df['월'] = df['측정일시'].dt.month

# 월별로 그룹화하여 미세먼지와 초미세먼지의 평균을 계산합니다.
monthly_avg = df.groupby('월')[['미세먼지(PM10)', '초미세먼지(PM2.5)']].mean().reset_index()
print(monthly_avg)

# 월별 평균 농도를 막대 그래프로 시각화합니다.
plt.figure(figsize=(12, 6))
sns.barplot(data=monthly_avg, x='월', y='미세먼지(PM10)', color='skyblue', label='미세먼지(PM10)')
sns.barplot(data=monthly_avg, x='월', y='초미세먼지(PM2.5)', color='lightgreen', label='초미세먼지(PM2.5)')
plt.title('월별 평균 미세먼지 및 초미세먼지 농도')
plt.xlabel('월')
plt.ylabel('평균 농도 (µg/m³)')
plt.legend()
plt.show()


# --- 6. 미세먼지-초미세먼지 상관관계 분석 및 히트맵 (추가 기능 2) ---
print("\n[미세먼지-초미세먼지 상관관계 분석]")

# 상관관계 분석에 사용할 컬럼만 선택합니다.
corr_df = df[['미세먼지(PM10)', '초미세먼지(PM2.5)']]

# 상관계수 행렬을 계산합니다.
correlation_matrix = corr_df.corr()
print("상관계수 행렬:")
print(correlation_matrix)

# 히트맵으로 상관관계를 시각화합니다.
plt.figure(figsize=(6, 5))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('미세먼지(PM10)와 초미세먼지(PM2.5)의 상관관계')
plt.show()
```
